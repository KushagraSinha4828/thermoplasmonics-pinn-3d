```python
import matplotlib.pyplot as plt
from nanothermal import NanoThermalSolver # Adjust import based on your exact structure

# 1. Update matplotlib parameters globally for BIGGER FONTS
plt.rcParams.update({
    'font.size': 14,
    'axes.titlesize': 18,
    'axes.labelsize': 16,
    'xtick.labelsize': 14,
    'ytick.labelsize': 14,
    'legend.fontsize': 14,
    'figure.titlesize': 22
})

# Define the path to your TDMS file
tdms_path = r"path_to_your_file.tdms"

print("Running Automated Pipeline Diagnostics...")

# 2. Execute the Automated Cleaning Pipeline
# The show_pipeline=True argument natively generates the 12-panel plot
# Setting noise_specs=None triggers the automated radial profile noise detection
cleaned_opd_nm = NanoThermalSolver.process_and_clean_tdms(
    tdms_path=tdms_path,
    noise_specs=None,              # Triggers Automated Mode[cite: 1]
    show_pipeline=True,            # <-- THIS GENERATES THE 12-PANEL PLOT[cite: 1]
    sigma_threshold=3.0,           # Controls sensitivity of the outlier detection[cite: 1]
    dc_exclusion_radius_px=2,      # Protects the DC core[cite: 1]
    sb_exclusion_radius_px=45,     # Protects the sideband[cite: 1]
    probe_lambda_nm=457.0,         # Automatically returns true OPD (nm)[cite: 1]
    spot_radius_um=4               # Masking radius for auto-detected spots[cite: 1]
)

# 3. (Optional) Plot the Final Cleaned Phase Map individually with large, bold labels
plt.figure(figsize=(10, 8))
im = plt.imshow(cleaned_opd_nm, cmap='turbo')

plt.title("Final Cleaned OPD Map (nm)", fontsize=24, fontweight='bold', pad=15)
plt.xlabel("X (pixels)", fontsize=18, fontweight='bold')
plt.ylabel("Y (pixels)", fontsize=18, fontweight='bold')

cbar = plt.colorbar(im)
cbar.set_label("Optical Path Difference (nm)", fontsize=18, fontweight='bold')
cbar.ax.tick_params(labelsize=14)

plt.tight_layout()
plt.show()
```
